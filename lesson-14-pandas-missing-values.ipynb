# Handle Missing Data with pandas

Make an explicit choice about whitespace, then compare dropping and filling values.

```python
import pandas as pd

data = pd.DataFrame({"mpg": [18.0, None, 24.0, 32.0], "horsepower": [100.0, 90.0, None, 72.0], "origin": ["USA", " ", "Europe", "Japan"]})
assert data["origin"].isna().sum() == 0
clean = data.replace(r"^\s*$", pd.NA, regex=True)
assert clean.isna().sum().to_dict() == {"mpg": 1, "horsepower": 1, "origin": 1}
print(clean.isna().sum())
```

```python
complete_measurements = clean.dropna(subset=["mpg", "horsepower"])
assert len(complete_measurements) == 2
filled = clean.fillna({"horsepower": clean["horsepower"].median(), "origin": "Unknown"})
assert filled.loc[2, "horsepower"] == 90
assert filled["mpg"].isna().sum() == 1
print(filled)
```
