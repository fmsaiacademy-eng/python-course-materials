## Compare arrays and lists

Multiplying a list by an integer repeats the list. Multiplying a numeric array by an integer scales every element.

```python
numbers = [1, 2, 3]
array = np.array(numbers)
assert numbers * 2 == [1, 2, 3, 1, 2, 3]
assert (array * 2).tolist() == [2, 4, 6]
```

## Inspect dimensions and index an array

Use `ndim` for the number of axes, `shape` for axis lengths, `size` for the total number of elements, and `dtype` for the element type.

```python
a = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.int64)
assert a.ndim == 2 and a.shape == (2, 3) and a.size == 6
assert a[1, 2] == 6
assert a[:, 0].tolist() == [1, 4]
```

## Build values with zeros, arange, and linspace

`arange` uses a step and normally excludes the stop. `linspace` uses a number of samples and includes the endpoint by default.

```python
assert np.zeros((2, 3), dtype=int).tolist() == [[0, 0, 0], [0, 0, 0]]
assert np.arange(2, 10, 2).tolist() == [2, 4, 6, 8]
assert np.linspace(2, 10, 5).tolist() == [2, 4, 6, 8, 10]
```

## Reshape, transpose, and operate on arrays

A reshape preserves the total element count. Transposing a 2D array switches rows and columns.

```python
a = np.arange(1, 7).reshape(2, 3)
assert a.T.shape == (3, 2)
assert a.T.tolist() == [[1, 4], [2, 5], [3, 6]]
assert (a + 10)[0, 0] == 11
assert a.max() == 6
assert a.reshape(3, 2).size == a.size
```

## Compare a copy and a view

A view can share the original array's data. A copy holds independent data.

```python
original = np.array([1, 2, 3])
view = original.view()
copied = original.copy()
view[0] = 99
assert original[0] == 99
assert copied[0] == 1
```

## Concatenate, stack, and flatten

Concatenate joins along an existing axis. Stacking adds an axis or arranges compatible arrays in rows or columns, depending on the function. Flatten makes a 1D copy.

```python
first, second = np.array([1, 2]), np.array([3, 4])
assert np.concatenate([first, second]).tolist() == [1, 2, 3, 4]
stacked = np.vstack([first, second])
assert stacked.tolist() == [[1, 2], [3, 4]]
assert stacked.flatten().tolist() == [1, 2, 3, 4]
```

## Generate reproducible values and change the dtype

Use a seed with `default_rng` for a reproducible sequence. Use `astype` to convert the element type.

```python
rng = np.random.default_rng(42)
sample = rng.integers(0, 10, size=(2, 3))
assert np.array_equal(sample, np.random.default_rng(42).integers(0, 10, size=(2, 3)))
assert sample.astype(float).dtype == np.dtype(float)
```
