# Combine DataFrames with pandas

Use concat to stack objects, merge to match column keys, and join for an index-oriented workflow.

```python
import pandas as pd
```

```python
jan = pd.DataFrame({"id": [1, 2], "sales": [120, 80]})
feb = pd.DataFrame({"id": [3], "region": ["East"]})
combined = pd.concat([jan, feb], ignore_index=True)
assert combined.shape == (3, 3)
assert combined["sales"].isna().sum() == 1
print(combined)
```

```python
customers = pd.DataFrame({"id": [1, 2, 3], "name": ["Asha", "Dev", "Ira"]})
orders = pd.DataFrame({"id": [1, 1, 2, 4], "amount": [60, 70, 30, 90]})
for how, size in {"inner": 3, "left": 4, "right": 4, "outer": 5}.items():
    result = customers.merge(orders, on="id", how=how, validate="one_to_many")
    assert len(result) == size
    print(how, len(result))
```

```python
indexed = customers.set_index("id").join(orders.set_index("id"), how="left")
assert indexed.index.tolist() == [1, 1, 2, 3]
print(indexed)
```
