## One axis and two axes

A Series has an index. A DataFrame has an index and columns, even when it has a single column.

```python
array_1d = np.array([10, 20, 30])
array_2d = array_1d[:, None]
series = pd.Series(array_1d, name="score")
frame = series.to_frame()
assert array_1d.shape == series.shape == (3,)
assert array_2d.shape == frame.shape == (3, 1)
assert isinstance(frame["score"], pd.Series)
assert isinstance(frame[["score"]], pd.DataFrame)
print(series, frame, sep="\n\n")
```

## Read and inspect a CSV

For a real CSV, pass its file path to `pd.read_csv`. This example keeps its input self-contained.

```python
sample = io.StringIO("city,score\nBengaluru,10\nHyderabad,NA\nKurnool,30\n")
data = pd.read_csv(sample)
print(data.shape)
data.info(show_counts=True)
print(data.isna().sum())
assert data.isna().sum().to_dict() == {"city": 0, "score": 1}
```

```python
print(data.head(2))
print(data.tail(2))
assert len(data.head()) == min(len(data), 5)
```
