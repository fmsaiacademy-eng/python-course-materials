## Group even, odd, and prime numbers

A prime has exactly two positive divisors. A composite integer has a divisor at or below its integer square root. Prime numbers can also be even or odd, so the prime test is a separate check.

```python
from math import isqrt

def is_prime(number):
    if number < 2:
        return False
    for divisor in range(2, isqrt(number) + 1):
        if number % divisor == 0:
            return False
    return True

def classify_numbers(numbers):
    evens, odds, primes = [], [], []
    for number in numbers:
        if number % 2 == 0:
            evens.append(number)
        else:
            odds.append(number)
        if is_prime(number):
            primes.append(number)
    return evens, odds, primes

evens, odds, primes = classify_numbers([-3, 0, 1, 2, 3, 4, 9, 11, 25])
assert evens == [0, 2, 4]
assert odds == [-3, 1, 3, 9, 11, 25]
assert primes == [2, 3, 11]
assert not is_prime(1) and not is_prime(49)
```

## Remove duplicates

Converting to a set removes duplicates. Use a loop with a set of seen values to preserve first-seen order.

```python
numbers = [3, 1, 3, 2, 1, 4, 2]
assert set(numbers) == {1, 2, 3, 4}

seen = set()
unique = []
for number in numbers:
    if number not in seen:
        seen.add(number)
        unique.append(number)

assert unique == [3, 1, 2, 4]
print(unique)
```

## Create, read, and update a dictionary

Use a hashable key such as a string. Square brackets look up a key, and `get` can return a default for a missing key.

```python
prices = {"pen": 10, "notebook": 25}
assert prices["pen"] == 10
prices["pen"] = 12
prices["eraser"] = 5
assert prices.get("pencil", 0) == 0
assert list(prices.keys()) == ["pen", "notebook", "eraser"]
assert list(prices.values()) == [12, 25, 5]
assert list(prices.items())[0] == ("pen", 12)
```

## Access nested values

A dictionary value can itself be a dictionary or a list. Chain lookups to reach an inner value.

```python
inventory = {"stationery": {"pens": [8, 12, 5], "notebooks": 25}}
assert inventory["stationery"]["pens"][0] == 8
inventory["stationery"]["notebooks"] -= 1
assert inventory["stationery"]["notebooks"] == 24
```

## Check repeated keys and repeated values

A repeated key in a literal has one final value. Different keys can share the same value.

```python
latest = {"pen": 10, "pen": 12}
assert latest == {"pen": 12}
assert list({"pen": 12, "pencil": 12}.values()) == [12, 12]
```
