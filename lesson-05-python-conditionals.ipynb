## Compare two values with `if` and `else`

```python
person1, person2 = 30, 50

if person1 > person2:
    print("Person 1 has the higher weight.")
else:
    print("Person 1 does not have the higher weight.")
```

## Return a result from a function

Use `elif` for the second comparison and `else` for a tie.

```python
def compare_weights(first, second):
    if first > second:
        return "Person 1 has the higher weight."
    elif second > first:
        return "Person 2 has the higher weight."
    else:
        return "The weights are equal."

print(compare_weights(person1, person2))
print(compare_weights(50, 50))
```

## Compare three values with `and`

A branch wins only when its value is greater than both other values. The final branch handles ties for the highest weight.

```python
def heaviest_person(p1, p2, p3):
    if p1 > p2 and p1 > p3:
        return "Person 1"
    elif p2 > p1 and p2 > p3:
        return "Person 2"
    elif p3 > p1 and p3 > p2:
        return "Person 3"
    else:
        return "Tie for highest weight"

print(heaviest_person(30, 50, 70))
print(heaviest_person(50, 50, 30))
```

## Fix a `NameError`

`person1` and `p1` are different names. Pass the defined variables into the function. Python binds them to its parameters.

```python
person1, person2, person3 = 30, 50, 70
print(heaviest_person(person1, person2, person3))
```
