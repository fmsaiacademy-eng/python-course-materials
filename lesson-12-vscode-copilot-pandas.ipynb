## Verify a suggested Python function

Ask your coding assistant for a function that detects a palindrome while ignoring case and non-alphanumeric characters. Make the empty-input behavior explicit and test the proposed implementation.

```python
def is_palindrome(value):
    normalized = "".join(ch for ch in value.casefold() if ch.isalnum())
    return bool(normalized) and normalized == normalized[::-1]

assert is_palindrome("Madam")
assert is_palindrome("A man, a plan, a canal: Panama!")
assert not is_palindrome("Python")
assert not is_palindrome("")
```

## Review and check a pandas example

Compare a Series with a one-column DataFrame, run a simple aggregation, and check the expected result.

```python
import pandas as pd

values = pd.Series([4, 7, 9], name="score")
table = values.to_frame()
assert values.ndim == 1
assert table.ndim == 2
assert table.shape == (3, 1)
assert table["score"].sum() == 20
print(table)
```
