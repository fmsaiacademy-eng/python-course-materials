# Pandas Series, DataFrames and GroupBy

An offline practice notebook for Python Lesson 16. Install pandas and matplotlib in the notebook's Python environment.

```python
import pandas as pd

scores = pd.Series([10, 20, 30, 10])
print(scores * 2)
print(scores.sort_values())
print(scores.rank(method="average"))
assert scores.rank().tolist() == [1.5, 3.0, 4.0, 1.5]
names = pd.Series([" sam ", "CHARLIE", "amy"])
print(names.str.strip().str.title())
```

```python
ages = pd.Series([22, 25, 30, None], dtype="Int64")
print(ages.agg(["sum", "count", "mean", "min", "max"]))
assert abs(ages.mean() - 77 / 3) < 1e-12
print(ages.fillna(22))
```

## Group and summarise sales

Each company and first name is an illustrative classroom example.

```python
from pathlib import Path

sales = pd.DataFrame({
    "company": ["Google", "Google", "Microsoft", "Microsoft", "Facebook", "Facebook"],
    "person": ["Sam", "Charlie", "Amy", "Vanessa", "Carol", "Sarah"],
    "sales": [200, 120, 340, 124, 243, 350],
})
by_company = sales.groupby("company")["sales"]
summary = by_company.agg(["sum", "mean", "max"])
print(summary)
top_rows = sales.loc[by_company.idxmax()].reset_index(drop=True)
print(top_rows)
assert summary["sum"].idxmax() == "Facebook"
top_rows.to_csv(Path("top_sales_by_company.csv"), index=False)
```

```python
import matplotlib.pyplot as plt

axis = summary["sum"].plot(kind="bar", title="Total sales per company", rot=0)
axis.set_ylabel("Sales")
plt.tight_layout()
plt.show()
```

## Scope and exception handling

A local variable can share a name with a module-level variable. Handle invalid text where you can respond to it. Here, the caller handles the expected ValueError and shows a helpful message.

```python
label = "module"

def local_label():
    label = "function"
    return label

assert local_label() == "function"
assert label == "module"

def is_even(number):
    return number % 2 == 0

for raw_number in ["10", "ten"]:
    try:
        number = int(raw_number)
    except ValueError:
        print(f"{raw_number!r} is not a valid whole number.")
    else:
        print(f"Is {number} even? {is_even(number)}")
```
