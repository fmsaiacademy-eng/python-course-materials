## Iterate over a string with `for`

Each iteration assigns the next character to the loop variable.

```python
phrase = "total data science"
for character in phrase:
    print(repr(character))
```

## Filter a list with a condition

```python
numbers = [1, 7, 3, 5, 2, 9, 66, 4]
greater_than_three = []
for number in numbers:
    if number > 3:
        greater_than_three.append(number)

print(greater_than_three)
assert greater_than_three == [7, 5, 9, 66, 4]
```

## Repeat with `while`

The loop checks its condition before each iteration. Update the counter inside the loop so the condition eventually becomes false.

```python
counter = 1
limit = 5
visited = []
while counter <= limit:
    visited.append(counter)
    counter += 1

print(visited)
assert visited == [1, 2, 3, 4, 5]
```

## Stop with `break`

```python
before_first_a = []
for character in phrase:
    if character == "a":
        break
    before_first_a.append(character)

print("".join(before_first_a))
assert "".join(before_first_a) == "tot"
```

## Skip one iteration with `continue`

```python
without_a = []
for character in phrase:
    if character == "a":
        continue
    without_a.append(character)

print("".join(without_a))
assert "".join(without_a) == phrase.replace("a", "")
```

## Practice · Print 1 through 10, skipping 8

Try it first, then compare your result with this solution.

```python
result = []
for number in range(1, 11):
    if number == 8:
        continue
    result.append(number)

print(*result)
assert result == [1, 2, 3, 4, 5, 6, 7, 9, 10]
```
