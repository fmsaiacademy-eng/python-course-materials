## Create, index, and slice lists

Indexing begins at zero. A slice includes its start and stops before its end.

```python
values = [1, 2, 4, 6, 8, 10]
mixed = ["pen", 3, 2.5, True]
print(mixed[0], values[-1], values[1:4])
assert values[1:4] == [2, 4, 6]
assert values[::-1] == [10, 8, 6, 4, 2, 1]
```

## Filter even and odd integers

The remainder operator `%` tests divisibility by 2. Return the two results as a tuple, then unpack it.

```python
def split_even_odd(numbers):
    evens, odds = [], []
    for number in numbers:
        if number % 2 == 0:
            evens.append(number)
        else:
            odds.append(number)
    return evens, odds

evens, odds = split_even_odd([0, 1, 2, 4, 5, -3, -2])
assert evens == [0, 2, 4, -2]
assert odds == [1, 5, -3]
print(evens, odds)
```

## Try list operators and methods

`+` creates a new list from two lists. `append`, `pop`, `sort`, and `reverse` modify the existing list.

```python
values = [3, 1] + [2]
assert [1, 2] * 2 == [1, 2, 1, 2]
values.append(4)
assert values.pop() == 4
values.sort()
assert values == [1, 2, 3]
values.reverse()
assert values == [3, 2, 1]
```

## Index into a nested list

The first index selects an inner list. The second index selects an item inside it.

```python
grid = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
assert grid[1][2] == 6
first_column = [row[0] for row in grid]
assert first_column == [1, 4, 7]
```

## Write a list comprehension

The expression before `for` gives each output value. The optional `if` controls which items are included.

```python
numbers = [1, 2, 3, 4, 5]
doubled = [2 * number for number in numbers]
even_numbers = [number for number in numbers if number % 2 == 0]
assert doubled == [2, 4, 6, 8, 10]
assert even_numbers == [2, 4]
```

## Compare lists, tuples, sets, and dictionaries

Lists and tuples preserve order and can contain repeated values. Tuple elements cannot be reassigned. Sets retain unique elements, and dictionaries map keys to values.

```python
items = ["pen", "pen", "notebook"]
fixed_items = tuple(items)
assert fixed_items.count("pen") == 2
assert fixed_items.index("notebook") == 2
assert set(items) == {"pen", "notebook"}
prices = {"pen": 10, "notebook": 25}
assert prices["pen"] == 10
```
