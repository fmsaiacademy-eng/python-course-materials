## Run the shorthand

Start with an integer and print it after adding 10.

```python
a = 5
a += 10
print(a)
assert a == 15
```

## Run the expanded form

Reset the starting value before comparing the two expressions.

```python
a = 5
a = a + 10
print(a)
assert a == 15
```

## Trace a sequence

Predict the result after each line, then run the cell to check.

```python
a = 5
a += 10
first_result = a
a += 10
assert (first_result, a) == (15, 25)
print(first_result, a)
```
